```python

from pathlib import Path
import pandas as pd
import os
```

```python

os.makedirs("data/ingestion", exist_ok=True)
data = "secom"
output ="data"
# Load raw sensor measurements
X = pd.read_csv(
    data+ "/secom.data",
    sep=r"\s+",
    header=None,
    na_values=["NaN"],
)

# Load labels and timestamps
labels = pd.read_csv(
    data+"/secom_labels.data",
    sep=r"\s+",
    header=None,
    names=["label_raw", "timestamp"],
)
```

```python
print("=== DATASET PROFILE ===")
print(f"Sensor records: {X.shape[0]:,}")
print(f"Sensor features: {X.shape[1]:,}")
print(f"Label records: {labels.shape[0]:,}")
print(f"Label columns: {labels.shape[1]}")
```

```python

# Assign source row IDs to preserve correspondence
X.columns = [
    f"sensor_{i:03d}" for i in range(1, X.shape[1] + 1)
]

df = pd.concat(
    [
        pd.Series(range(1, len(X) + 1), name="source_row_id"),
        X.reset_index(drop=True),
        labels.reset_index(drop=True),
    ],
    axis=1,
)

# Save without transforming labels or timestamps
output_file = output+ "/secom_raw_combined.csv"
df.to_csv(output_file, index=False)

print("Output shape:", df.shape)
print("Saved to:", output_file)
print(df[["source_row_id", "label_raw", "timestamp"]].head())
```
